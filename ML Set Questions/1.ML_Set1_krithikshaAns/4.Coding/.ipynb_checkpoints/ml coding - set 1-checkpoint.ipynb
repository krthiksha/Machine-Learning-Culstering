```python
#1.
Question 1:A retail store wants to identify customers who make frequent purchases. Given the dataset below, write a Python program to:
Group customers by their IDs.


Calculate the total purchase amount per customer.


Identify the top 3 customers with the highest purchase amounts.


Dataset:
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

Expected Output:
Total Purchases per Customer:
   Customer_ID  Purchase_Amount
0          101              520
1          102              600
2          103              480
3          104              300
4          105              400

Top 3 Frequent Customers:
   Customer_ID  Purchase_Amount
1          102              600
0          101              520
2          103              480
```

```python
import pandas as pd
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

""" USING PYTHON DICT AND PANDAS DATAFRAME   ---- APPROACH 1"""
# customer_id =data['Customer_ID']
# purchase_amount = data['Purchase_Amount'] 
# # manually looping to calculate the total purchase amount
# total_purchase = {}
# for i in range(len(customer_id)):
#     cus_id = customer_id[i]
#     if cus_id in total_purchase:
#         total_purchase[cus_id] += purchase_amount[i]
#     else:
#         total_purchase[cus_id] = purchase_amount[i]
# print(total_purchase)
# # convert dict into dataframe and sort the total_purchase in descending - to get the TOP 3 CUTOMERS 
# df = pd.DataFrame(list(total_purchase.items()), columns=['customer_id','total_purchase'])
# df = df.sort_values(by='total_purchase',ascending=False)
# df[:3]


"""USING PANDAS DATAFRAME - groupby --- APPROACH 2  (OPTIMIZED CODE)"""
# converting data into dataframe
data_df = pd.DataFrame(data)
# use groupby - to calculate the total purchase amount for each customer ID 
grouped_data = data_df.groupby('Customer_ID',as_index=False)['Purchase_Amount'].sum()
print("\nCUSTOMERS WITH TOTAL PURCHASE: \n", grouped_data)
# sort the top 3 customers using sort values in pandas
sorted_data = grouped_data.sort_values('Purchase_Amount',ascending=False)
print("\nTOP 3 CUSTOMERS: \n",sorted_data[:3])
```

```python
#2. Predicting House Prices with Linear Regression
A real estate company wants to predict house prices based on square footage. Write a Python program to:
Train a Linear Regression model.


Predict house prices for given test data.


Dataset:
data = {'Square_Feet': [1500, 2000, 2500, 3000, 3500],
        'Price': [300000, 400000, 500000, 600000, 700000]}

Test Data: [[1800], [2800]]
Expected Output:
Predicted Prices:
[360000. 560000.]

```

ANSWER:-

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
# load data
data = {'Square_Feet': [1500, 2000, 2500, 3000, 3500],
        'Price': [300000, 400000, 500000, 600000, 700000]}
# split input and output
x=pd.DataFrame(data['Square_Feet'])
y=pd.DataFrame(data['Price'])
# split train and test dataset
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.30, random_state=0)
# print(x_train, x_test, y_train, y_test )
# standardize input - x
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train =sc.fit_transform(x_train)
x_test = sc.transform(x_test)
# model creation 
model_regressor = LinearRegression()
model_regressor.fit(x_train,y_train)
# test the model 
y_predicted = model_regressor.predict(x_test)
print("y_predicted: ",y_predicted)
# model evaluation 
from sklearn.metrics import r2_score
r2_score = r2_score(y_test, y_predicted)
print("r2_score: ", r2_score)

# testing user input data in model 
Test_Data = [[1800], [2800]]
Test_Data=sc.transform(Test_Data)
# prediction 
house_price = model_regressor.predict(Test_Data)
print("house_price: ", house_price)
"""
model returned output: 
 [[360000.]
 [560000.]]

 r2_score:  1.0  => PERFECT REGRESSION MODEL 
 """

```

```python
#3:
Identifying Frequent Labels in a Dataset
A company wants to identify the top 3 most common categories in a dataset. Given the dataset below, write a Python program to:
Group the data by Category.


Count the total occurrences of each category.


Identify the top 3 most frequent categories.


Dataset:
data = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}

Expected Output:

Total Occurrences per Category:
  Category  Count
0        A      4
1        B      3
2        C      3
3        D      1
4        E      1

Top 3 Frequent Categories:
  Category  Count
0        A      4
1        B      3
2        C      3
```

ANSWER:-

```python
import pandas as pd
data = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}
# dataframe conversion
data = pd.DataFrame(data)
# Total Occurrences per Category:
data = data.groupby('Category',as_index=False)['Value'].sum()
print("\nTotal Occurrences per Category: \n",data)
# Top 3 Frequent Categories:
data = data.sort_values('Value',ascending=False)
print("\nSorted Categories: \n",data)
print("\nTop 3 Frequent Categories: \n",data[:3])

```

```python
#4.
Predicting Missing Values Using Mean Imputation
A dataset contains missing values in the Age column. Write a Python program to:
Replace missing values with the mean of the column.


Display the updated DataFrame.


Dataset:
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, None, 35, None]}

Expected Output:

Original Data:
     Name   Age
0  Alice  25.0
1    Bob  30.0
2 Charlie   NaN
3  David  35.0
4    Eve   NaN

Data after Imputation:
     Name   Age
0  Alice  25.0
1    Bob  30.0
2 Charlie  30.0
3  David  35.0
4    Eve  30.0
```

ANSWER:-

```python
import pandas as pd 
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, None, 35, None]}
# convert dataframe
data=pd.DataFrame(data)
print("Original Data: \n", data)
# find mean
mean = data['Age'].mean()
# Replace missing values with the mean of the column.
data.fillna(value=mean, inplace=True)
print("\nData after Imputation: \n",data)
```

```python
# 5: 
Implementing a Simple Linear Regression Model
You are given a dataset with Experience (years) and Salary ($). Write a Python program to:
Train a Linear Regression model.


Predict the salary for an individual with 6 years of experience.


Dataset:
import pandas as pd

data = {'Experience': [1, 2, 3, 4, 5], 
        'Salary': [30000, 35000, 40000, 45000, 50000]}
df = pd.DataFrame(data)

Expected Output (Example):

Predicted Salary for 6 years of experience: $55000


data = {'word_count': [100, 150, 200, 120, 180, 220], 
        'is_spam': ['ham', 'spam', 'spam', 'ham', 'spam', 'spam']}
df = pd.DataFrame(data)

Expected Output (Example):
Prediction for email with 200 words: Spam

```

ANSWER:-

```python
# REGRESSION PROBLEM 
from sklearn.linear_model import LinearRegression
import pandas as pd
# load data
data = {'Experience': [1, 2, 3, 4, 5], 
        'Salary': [30000, 35000, 40000, 45000, 50000]}
# split input and output
x=pd.DataFrame(data['Experience'])
y=pd.DataFrame(data['Salary'])
# split train and test dataset
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.30, random_state=0)
# print(x_train, x_test, y_train, y_test )
# standardize input - x
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train =sc.fit_transform(x_train)
x_test = sc.transform(x_test)
# model creation 
model_regressor = LinearRegression()
model_regressor.fit(x_train,y_train)
# test the model 
y_predicted = model_regressor.predict(x_test)
print("y_predicted: ",y_predicted)
# model evaluation 
from sklearn.metrics import r2_score
r2_score = r2_score(y_test, y_predicted)
print("r2_score: ", r2_score)

# testing user input data in model 
Test_Data = [[6]]
Test_Data=sc.transform(Test_Data)
# prediction 
emp_salary = model_regressor.predict(Test_Data)
print("emp_salary: ", emp_salary)
"""
model returned output: 
 r2_score:  1.0     => PERFECT REGRESSION MODEL 
emp_salary:  [[55000.]]
 """
```

```python
# CLASSIFICATION PROBLEM 
import pandas as pd 
data = {'word_count': [100, 150, 200, 120, 180, 220], 
        'is_spam': ['ham', 'spam', 'spam', 'ham', 'spam', 'spam']}
# convert categorical into numerical 
is_spam = pd.DataFrame(data['is_spam'])
is_spam = pd.get_dummies(is_spam, drop_first=True, dtype=int)
# split input & output 
independent = pd.DataFrame(data['word_count'])
dependent = is_spam
# split train & test dataset
from sklearn.model_selection import train_test_split
x_train,x_test, y_train,y_test = train_test_split(independent, dependent, test_size=0.30,random_state=0)
y_train = y_train.values.ravel()
# model creation 
from sklearn.linear_model import LogisticRegression 
classifier_model = LogisticRegression()
classifier_model.fit(x_train, y_train)
# test the model
y_pred = classifier_model.predict(x_test)
y_pred
# model evaluation
from sklearn.metrics import confusion_matrix, classification_report
confusion_matrix = confusion_matrix(y_test,y_pred, labels=[0,1])
print(confusion_matrix)
classification_report = classification_report(y_test,y_pred)
print(classification_report)

# USER INPUT _ TEST 
""" Expected Output (Example):
Prediction for email with 200 words: Spam"""
predicted_output = classifier_model.predict([[200]])
predicted_output
```
